```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('old_car_price_prediction.csv')
print(df.head())
print(df.info())
print(df.dtypes)
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.describe())
```

```python
print(df.columns.tolist())
```

```python
Car_name_le = LabelEncoder()
df['Car_name']= Car_name_le.fit_transform(df['Car_name'])
df.head(2)
```

```python
df['Fuel_Type'].unique()
```

```python
Fuel_Type_le = LabelEncoder()
df['Fuel_Type']= Fuel_Type_le.fit_transform(df['Fuel_Type'])
df.head(2)
```

```python
df['Seller_Type'].unique()
```

```python
Seller_Type_le = LabelEncoder()
df['Seller_Type']= Seller_Type_le.fit_transform(df['Seller_Type'])
df.head(2)
```

```python
df['Transmission'].unique()
```

```python
Transmission_le = LabelEncoder()
df['Transmission']= Transmission_le.fit_transform(df['Transmission'])
df.head(2)
```

```python
x= df[['Car_name','Year','KM_Driven','Fuel_Type','Seller_Type','Transmission','Owner','Present_Price_Lakh']]

y=df['Selling_Price_Lakh']
```

```python
ss = StandardScaler()
x = pd.DataFrame(ss.fit_transform(x), columns=x.columns)
df.head(2)
print(x.head(2))
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train, y_train)
dt.score(x_train, y_train)*100,dt.score(x_test, y_test)*100
```

```python
mean_absolute_error(y_test, dt.predict(x_test)),mean_squared_error(y_test, dt.predict(x_test)),r2_score(y_test, dt.predict(x_test))
```

```python
from sklearn.ensemble import RandomForestRegressor
ml = RandomForestRegressor()
ml.fit(x_train, y_train)
ml.score(x_train, y_train)*100, ml.score(x_test,y_test)*100
```

```python
mean_absolute_error(y_test, ml.predict(x_test)),mean_squared_error(y_test, ml.predict(x_test)),r2_score(y_test, ml.predict(x_test))
```

```python


joblib.dump(dt, "decision_tree.pkl")
joblib.dump(ml, "random_forest.pkl")

print("Both models saved successfully!")
```
